```python
# Import librarynye
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load dataset (anggaplah 1000 siswa data set nya wkwkwk)
data = pd.read_csv('nilai_siswa.csv')
print("📊 Data Preview:")
print(data)

# Hitung nilai akhir (anggap lah: 40% Tugas + 60% Ujian)
data['Nilai_Akhir'] = (0.4 * data['Tugas']) + (0.6 * data['Ujian'])
print("\n✅ Data with Nilai Akhir:")
print(data)

# Visualisasi data
plt.scatter(data['Tugas'], data['Nilai_Akhir'], color='blue', label='Tugas')
plt.scatter(data['Ujian'], data['Nilai_Akhir'], color='green', label='Ujian')
plt.xlabel('Nilai')
plt.ylabel('Nilai Akhir')
plt.legend()
plt.title('Distribusi Nilai')
plt.grid(True)
plt.show()

# Split data ke X (fitur) dan y (target)
X = data[['Tugas', 'Ujian']]
y = data['Nilai_Akhir']

# Split ke train dan test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

# Buat model Linear Regression
model = LinearRegression()
model.fit(X_train, y_train)

# Prediksi nilai test
y_pred = model.predict(X_test)

# Evaluasi model
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print(f"\n📊 MAE: {mae:.2f}")
print(f"📊 MSE: {mse:.2f}")
print(f"📊 RMSE: {rmse:.2f}")
print(f"📊 R² Score: {r2:.2f}")

# Visualisasi hasil prediksi
plt.scatter(range(len(y_test)), y_test, color='blue', label='Nilai Asli')
plt.scatter(range(len(y_pred)), y_pred, color='red', label='Prediksi')
plt.plot(range(len(y_pred)), y_pred, color='red', linestyle='--')
plt.title('Hasil Prediksi vs Nilai Asli')
plt.legend()
plt.grid(True)
plt.show()

# Prediksi manual nilai siswa baru
nilai_baru = pd.DataFrame({
    'Tugas': [88],
    'Ujian': [92]
})

prediksi_nilai_akhir = model.predict(nilai_baru)
print(f"\n📌 Prediksi Nilai Akhir untuk siswa baru: {prediksi_nilai_akhir[0]:.2f}")
```
